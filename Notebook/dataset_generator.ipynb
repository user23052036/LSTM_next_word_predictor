```python
from itertools import product
```

```python
nouns = [
    "cat", "dog", "bird", "fish", "child",
    "robot", "teacher", "boy", "girl", "rabbit"
]

objects = [
    "book", "ball", "box", "toy", "note",
    "cup", "lamp", "chair", "table", "bag"
]

places = [
    "school", "home", "park", "room", "garden",
    "market", "library", "river", "street", "office"
]

adjectives = [
    "small", "big", "fast", "slow", "quiet",
    "bright", "clean", "warm", "cold", "safe"
]

verbs = [
    "run", "jump", "walk", "play", "sleep",
    "move", "learn", "work", "read", "write"
]

foods = [
    "rice", "bread", "fruit", "milk", "water",
    "tea", "juice", "soup", "salt", "cake"
]

pronouns = ["i", "you", "we", "they"]

dataset = []
```

```python
def add_unique(target_dataset, new_items):
    for item in new_items:
        item = " ".join(item.split())
        if item not in target_dataset:
            target_dataset.append(item)
```

```python
# 1) Description patterns
add_unique(dataset, [f"the {n} is {a}" for n, a in product(nouns, adjectives)])

# the second parameter of add_unique is a list of items to be added to the dataset. 
# The list is generated using a list comprehension that creates sentences in the format "the {noun} is {adjective}" 
# for each combination of nouns and adjectives. The product function from the itertools module is used to 
# create all possible combinations of nouns and adjectives.

# the product function takes two iterables (in this case, nouns and adjectives) and returns an iterator that produces 
# the Cartesian product of the input iterables. This means that it generates all possible pairs of items, the leftmost item from the 
# first iterable and the rightmost item from the second iterable. For example, 
# if nouns = ["cat", "dog"] and adjectives = ["small", "big"], the product function will generate the following pairs:
# ("cat", "small"), ("cat", "big"), ("dog", "small"), ("dog", "big").
```

```python
# 2) Capability patterns
add_unique(dataset, [f"the {n} can {v}" for n, v in product(nouns, verbs)])
```

```python
# 3) Location patterns
add_unique(dataset, [f"the {n} is in the {p}" for n, p in product(nouns, places)])
```

```python
# 4) Support / position patterns
add_unique(dataset, [f"the {n} is on the {o}" for n, o in product(nouns, objects)])
add_unique(dataset, [f"the {n} is under the {o}" for n, o in product(nouns, objects)])
```

```python
# 5) Relation patterns
add_unique(dataset, [
    f"the {n1} is near the {n2}"
    for n1, n2 in product(nouns, nouns)
    if n1 != n2
])
```

```python
# 6) Preference patterns
add_unique(dataset, [f"{p} like the {o}" for p, o in product(pronouns, objects)])
```

```python
# 7) Eating patterns
add_unique(dataset, [f"{p} eat {f}" for p, f in product(pronouns, foods)])
```

```python
# 8) Drinking patterns
add_unique(dataset, [f"{p} drink {f}" for p, f in product(pronouns, ["water", "milk", "tea", "juice", "soup"])])
```

```python
# 9) Simple state patterns
add_unique(dataset, [f"{p} are {a}" for p, a in product(["we", "they"], adjectives)])
add_unique(dataset, [f"i am {a}" for a in adjectives])
add_unique(dataset, [f"you are {a}" for a in adjectives])
```

```python
# 10) Compound patterns
add_unique(dataset, [f"the {n1} and the {n2} are {a}" for n1, n2 in product(nouns, nouns) if n1 != n2 for a in adjectives])
```

```python

# Keep exactly 500 sentences
dataset = dataset[:500]
assert len(dataset) == 500
```
